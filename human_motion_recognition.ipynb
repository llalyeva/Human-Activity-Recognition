```python
import kagglehub
import os
import pandas as pd
path = kagglehub.dataset_download("uciml/human-activity-recognition-with-smartphones")
print("Path to dataset files:", path)
os.listdir(path)

train_csv = pd.read_csv(path+"/train.csv")
test_csv = pd.read_csv(path+"/test.csv")
```

```python
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE


activity = train_csv["Activity"].unique()
class_distribution = train_csv["Activity"].value_counts()

class_distribution.plot(kind="bar") #not bad distribution
plt.xlabel("Activity type")
plt.ylabel("Number of samples")
plt.show()

#shrink dimensions
X = train_csv.drop(columns = ["Activity","subject"]) #only features vector
Y = train_csv["Activity"] #just type of acitivity

#time for standartization. feature vectors should have the same scale
scaler = StandardScaler() # x - mean/std

X_scaled = scaler.fit_transform(X)


tsne = TSNE(
    n_components=2,
    random_state=42,
    perplexity=30
)

X_tsne = tsne.fit_transform(X_scaled)


print("X.shape", X.shape)
print("Y.shape", Y.shape)
print("X.shape after tsne transform", X_tsne.shape)
```
